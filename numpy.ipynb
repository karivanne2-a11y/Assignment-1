# Assignment 1 – Parte 4: NumPy

Ejercicios básicos con arreglos y matrices de NumPy.

```python
import numpy as np
```

## 1. Crear un arreglo con los números del 1 al 10

```python
arr = np.arange(1, 11)
```

## 2. Imprimir el arreglo

```python
print("Arreglo:", arr)
```

## 3. Valor máximo y mínimo

```python
print("Máximo:", arr.max())
print("Mínimo:", arr.min())
```

## 4. Arreglo de 5 ceros

```python
zeros = np.zeros(5)
print("Ceros:", zeros)
```

## 5. Arreglo de 5 unos

```python
ones = np.ones(5)
print("Unos:", ones)
```

## 6. Crear la matriz

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

## 7. Mostrar la matriz

```python
print("Matriz:")
print(matrix)
```

## 8. Dimensiones con `.shape`

```python
print("Dimensiones (filas, columnas):", matrix.shape)
```
