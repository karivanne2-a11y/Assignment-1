# Assignment 1 – Parte 2: Tuplas

Ejercicios con la tupla `ages`.

## Crear la tupla

```python
ages = (20, 25, 22, 30, 28)
```

## 1. Imprimir la tupla

```python
print("Tupla:", ages)
```

## 2. Primer y último elemento

```python
print("Primer elemento:", ages[0])
print("Último elemento:", ages[-1])
```

## 3. Valor máximo, valor mínimo y número de elementos

```python
print("Máximo:", max(ages))
print("Mínimo:", min(ages))
print("Número de elementos:", len(ages))
```

## 4. Intentar modificar un elemento de la tupla

Al ejecutar la línea `ages[0] = 21` Python lanza un `TypeError`. Aquí lo capturamos con `try/except` para mostrar el error sin detener el notebook.

```python
try:
    ages[0] = 21
except TypeError as e:
    print("Error:", e)

# Otra opción: dejar la línea comentada una vez comprobado el error
# ages[0] = 21
```

## 5. ¿Por qué Python no permite esta modificación?

```python
# Python no permite modificar una tupla porque las tuplas son INMUTABLES:
# una vez creadas, sus elementos no pueden cambiarse, agregarse ni eliminarse.
# Esto garantiza que los datos se mantengan constantes y seguros durante
# la ejecución del programa (por ejemplo, para datos que no deben alterarse
# por error) y permite que las tuplas sean más eficientes y puedan usarse
# como claves de diccionarios.
# Si se necesita modificar los datos, se debe usar una lista.
print("Las tuplas son inmutables: no admiten asignación de elementos.")
```
