# Assignment 1 – Parte 1: Listas

Ejercicios con la lista `grades` usando `append()`, `sort()`, `max()`, `min()` y `len()`.

## Crear la lista

```python
grades = [15, 18, 12, 20, 16]
```

## 1. Imprimir la lista original

```python
print("Lista original:", grades)
```

## 2. Agregar la nota 17 con `append()`

```python
grades.append(17)
print("Después de append(17):", grades)
```

## 3. Ordenar de menor a mayor con `sort()`

```python
grades.sort()
print("Lista ordenada:", grades)
```

## 4. Mostrar la nota más alta, la más baja y el total de notas

```python
print("Nota más alta:", max(grades))
print("Nota más baja:", min(grades))
print("Total de notas:", len(grades))
```

## 5. Imprimir la lista final

```python
print("Lista final:", grades)
```
