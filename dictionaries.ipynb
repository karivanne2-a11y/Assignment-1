### Diccionario inicial: población de ciudades

```python
population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}
```

### 1. Imprimir el diccionario completo

```python
print(population)
```

### 2. Mostrar solo las claves con `keys()`

```python
print(population.keys())
```

### 3. Obtener la población de "Lima" con `get()`

```python
print(population.get("Lima"))
```

### 4. Agregar "Cusco": 428000

```python
population["Cusco"] = 428000
print(population)
```

### 5. Eliminar "Trujillo" con `pop()`

```python
population.pop("Trujillo")
print(population)
```

### 6. Imprimir el diccionario final

```python
print(population)
```
